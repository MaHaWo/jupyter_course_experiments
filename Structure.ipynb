In part A, we want to do 1), 2) and 3) and we are using technique a), b), c) and d) for it. We start out with 1) using technique a) and b):  

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from collections.abc import Callable
plt.rcParams['text.usetex'] = True

# use functions, type hints, docstrings, comments
def f(x: np.ndarray | float) -> np.ndarray | float:
    """
        Compute square of input

    ## Args:
    x (np.ndarray | float): input data. Either number or array of numbers

    ## Returns:
    x^2 (np.ndarray | float): The input squared, same type as input
    """
    return x**2

# make visualizations
def plot_xy(x: np.ndarray, y:  np.ndarray, label: str, title: str)-> plt.Figure:
    """
        Compute and plot f(x).
    ## Args:
    x (np.ndarray): input data
    y (np.ndarray): Oridnale data
    label (str): y-axis label
    title (str): figure title

    ## Returns:
    matplotlib.pyplot figure showing plotted data as graph y(x)
    """
    plt.plot(x, y)
    plt.title(title)
    plt.xlabel(r'$\displaystyle x $ [-]')
    plt.ylabel(label)
    return plt.gcf()
````

````python
import numpy as np
x = np.linspace(0, 10, 250)
y = f(x)
fig = plot_xy(x,y, label= r'$\displaystyle x^2$ [-]', title = "1) with a) and b): simple parabola")
````

### Part A with method c

Plot above shows that x,y,z are the case and u, v, w are not. This hints at 'something something'. Do 3) with c) now. The function below implements a 3rd order polynomial with offset 1, scaling factor 0.3 and quadratic term $$ 2 x^2 $$  

````python
# use functions, type hints, docstrings, comments
def g(x: np.ndarray | float) -> np.ndarray | float:
    """
        Compute thrid power of input
    x (np.ndarray | float): input data. Either number or array of numbers
    returns:
    x^3 (np.ndarray | float): The input squared, same type as input
    """
    return 0.3*x**3 - 2*x**2 + 1
````

````python
import numpy as np
x = np.linspace(0, 10, 250)
y = g(x)
fig = plot_xy(x,
              y,
              title = "3) with c): third order polynomial",
              label = r"$\displaystyle 0.3 x^3 - 2 x^2 + 1 $ [-]")
````
