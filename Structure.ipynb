In part A, we want to do 1), 2) and 3) and we are using techniques a), b), c) and d) for it. We start out with 1) using techniques a) and b):

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from collections.abc import Callable
plt.rcParams['text.usetex'] = True

# use functions, type hints, docstrings, comments
def f(x: np.ndarray | float) -> np.ndarray | float:
    """
        Compute square of input

    ## Args:
    x (np.ndarray | float): input data. Either number or array of numbers

    ## Returns:
    x^2 (np.ndarray | float): The input squared, same type as input
    """
    return x**2

# make visualizations
def plot_xy(x: np.ndarray, y: np.ndarray, label: str, title: str) -> plt.Figure:
    """
        Plot y over x.
    ## Args:
    x (np.ndarray): abscissa data
    y (np.ndarray): ordinate data
    label (str): y-axis label
    title (str): figure title

    ## Returns:
    matplotlib.pyplot figure showing plotted data as graph y(x)
    """
    plt.plot(x, y)
    plt.title(title)
    plt.xlabel(r'$\displaystyle x $ [-]')
    plt.ylabel(label)
    return plt.gcf()
````

````python
x = np.linspace(0, 10, 250)
y = f(x)
fig = plot_xy(x, y, label=r'$\displaystyle x^2$ [-]', title="1) with a) and b): simple parabola")
````

### Part A with method c

The plot above shows that x, y, z are the case and u, v, w are not. This hints at 'something something'. Do 3) with c) now. The function below implements a 3rd order polynomial with offset 1, scaling factor 0.3 and quadratic term $-2 x^2$.

````python
# use functions, type hints, docstrings, comments
def g(x: np.ndarray | float) -> np.ndarray | float:
    """
        Compute third order polynomial 0.3 x^3 - 2 x^2 + 1 of input

    ## Args:
    x (np.ndarray | float): input data. Either number or array of numbers

    ## Returns:
    0.3 x^3 - 2 x^2 + 1 (np.ndarray | float): The polynomial evaluated at the input, same type as input
    """
    return 0.3*x**3 - 2*x**2 + 1
````

````python
x = np.linspace(0, 10, 250)
y = g(x)
fig = plot_xy(x,
              y,
              title="3) with c): third order polynomial",
              label=r"$\displaystyle 0.3 x^3 - 2 x^2 + 1 $ [-]")
````

## Part B 

Train a simple MNIST MLP classifier with Lightning and PyTorch, using modules bundled in a package to support this notebook.

````python
import lightning as L
from lightning.pytorch.callbacks import TQDMProgressBar
from jupyter_ssc_compact_course.datasets import get_dataloaders
from jupyter_ssc_compact_course.model import SimpleNet
from jupyter_ssc_compact_course.training import LitClassifier
import torch 
torch.set_float32_matmul_precision('high')

train_loader, test_loader = get_dataloaders()
lit_model = LitClassifier(SimpleNet())
trainer = L.Trainer(max_epochs=3, callbacks=[TQDMProgressBar(refresh_rate=200)]) # limit message rate to get rid of annoying warnings 
trainer.fit(lit_model, train_loader)
trainer.test(lit_model, test_loader)
````
